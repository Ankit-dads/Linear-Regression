# multiclass classification

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# load dataset
iris = sns.load_dataset('iris')
iris.head()
```

```python
x = iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']]
y = iris['species']
```

```python
x.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(iris['species'])
```

```python
y
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.2,random_state=24)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model1 = LogisticRegression()
model1.fit(x_train,y_train)
```

```python
y_test
```

```python
y_pred = model1.predict(x_test)
y_pred
```

# check confusion matrix and accuracy score 

```python
from sklearn.metrics import accuracy_score,confusion_matrix
print(f"{confusion_matrix(y_test,y_pred)}")
print(f"{accuracy_score(y_test,y_pred)}")
```

# If we predict value which is not present in the original dataset

```python
result1 = model1.predict([[1,2,4,1]])
if result1[0] == 0: #result[0] we give [0] because only result() will give array but result[0] give integer value which can be compared easily
    print('The flower is Setosa.')
if result1[0] == 1:
    print('The Flower is Versicolor.')
else:
    print("The Flower is Virginica.")
```
