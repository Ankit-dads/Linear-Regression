# Performing EDA and predicting penguins species by using Logistic Regression.

* Multiclass Classification.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings # ignore warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv('penguins.csv')
df.tail()
```

```python
df.columns # it will give name of columns of dataset
```

```python
df.duplicated().sum() # data have no duplicate values in it.
```

```python
df.drop(columns= ['island','sex'],inplace = True)
df.head()
```

* changing body mass into kilograms(kg).

```python
df['body_mass_kg'] = df['body_mass_g']/1000
df.head()
```

```python
df.drop(columns= ['body_mass_g'],inplace =True)
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
df.isnull().sum()
```

* filing null(int/float) values with their mean values 

```python
df['bill_depth_mm'] = df['bill_depth_mm'].fillna(df['bill_depth_mm'].mean())
df['bill_depth_mm'].isnull().sum()
```

```python
df['bill_length_mm'] = df['bill_length_mm'].fillna(df['bill_length_mm'].mean())
df['flipper_length_mm'] = df['flipper_length_mm'].fillna(df['flipper_length_mm'].mean())
df['body_mass_kg'] = df['body_mass_kg'].fillna(df['body_mass_kg'].mean())
```

```python
df.isnull().sum()
```

# now the data is cleaned

```python
df.info()
```

```python
df.columns
```

```python
x = df[['bill_length_mm', 'bill_depth_mm','flipper_length_mm','body_mass_kg']]
y = df['species']
```

```python
x
```

```python
y
```

* transforming y feature into 0,1,2 format

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(y)
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=23)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

```python
y_test
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```

# now predict by giving some random values to it.

```python
result = model.predict([[45,14,212,65]])
if result[0] == 0:
    print('Penguin belongs to Adelie species')
elif result[0] == 1:
    print('Penguin belongs to Chinstrap Species')
else:
    print("Penguin belongs to Gentoo species")
```
