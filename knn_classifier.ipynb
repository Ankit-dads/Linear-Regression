# KNN CLASSIFIER (K NEAREST NEIGHBOUR)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

# load iris dataset froms seaborn to perform classification

```python
df = sns.load_dataset('iris')
df.head()
```

```python
x = df[['sepal_length','sepal_width','petal_length','petal_width']]
y = df['species']
```

```python
# standardization of inputs columns
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
print(x_scaled)
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y_encoded = le.fit_transform(y)
y_encoded
```

```python
fig = plt.figure(figsize=(5, 3))
sns.scatterplot(x=x_scaled[:, 0], y=x_scaled[:, 1], hue=y_encoded, palette="Set1")
plt.title("Scatter Plot of Scaled Features")
plt.xlabel("Sepal Length (scaled)")
plt.ylabel("Sepal Width (scaled)")
plt.legend(title="Species", labels=le.classes_)
plt.show()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x_scaled,y_encoded,test_size=0.2, random_state=24)
```

```python
parameters = {'n_neighbors': [3,5,7,9],}
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import RandomizedSearchCV
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn  = KNeighborsClassifier()
```

```python
grid = GridSearchCV(estimator=knn,param_grid=parameters,cv=5)
grid.fit(x_train,y_train)
print("Best Hyperparameters:", grid.best_params_)
```

```python
random = RandomizedSearchCV(estimator=knn , param_distributions=parameters, cv = 5)
random.fit(x_train,y_train)
print('Best Hyperparametere:', random.best_params_)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=7)
```

```python
knn
```

```python
knn.fit(x_train,y_train)
```

```python
y_pred = knn.predict(x_test)
y_pred
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
print('confusion matrix:\n', confusion_matrix(y_test,y_pred))
print('Accuracy score:', accuracy_score(y_test,y_pred))
```
