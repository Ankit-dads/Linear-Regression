```python
import seaborn as sns               # To load the tips dataset
import pandas as pd                 # To work with data tables (DataFrames)
import numpy as np                  # For math operations
import matplotlib.pyplot as plt     # To draw graphs

from sklearn.model_selection import train_test_split  # To split data
from sklearn.linear_model import LinearRegression      # Our ML model
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
# ↑ To check how good our model is performing
```

```python
# Load the dataset
tips = sns.load_dataset('tips')

# Peek at first 5 rows
print(tips.head())

# Basic info
print(tips.shape)   # (244, 7) → 244 rows, 7 columns
print(tips.info())
print(tips.describe())
```

```python
X = tips[['total_bill']]   # Feature (input) — 2D array needed for sklearn
y = tips['tip']             # Target (output) — what we want to predict

print(f"X shape: {X.shape}")  # (244, 1)
print(f"y shape: {y.shape}")  # (244,)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,      # 20% data for testing, 80% for training
    random_state=42     # Seed for reproducibility (same split every time)
)

print(f"Training samples: {X_train.shape[0]}")  # 195
print(f"Testing samples:  {X_test.shape[0]}")   # 49
```

```python
# Create the model
model = LinearRegression()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled  = scaler.transform(X_test)
```

```python
model.fit(X_train_scaled, y_train)
```

```python
y_pred = model.predict(X_test_scaled)
```

```python


# Compare actual vs predicted (first 5)
comparison = pd.DataFrame({'Actual Tip': y_test.values, 'Predicted Tip': y_pred.round(2)})
print(comparison.head(10))
```

```python
mae  = mean_absolute_error(y_test, y_pred)
mse  = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2   = r2_score(y_test, y_pred)

print(f"MAE  (Mean Absolute Error):  {mae:.4f}")
print(f"MSE  (Mean Squared Error):   {mse:.4f}")
print(f"RMSE (Root Mean Sq. Error):  {rmse:.4f}")
print(f"R²   (R-Squared Score):      {r2:.4f}")
```

```python
single = Sequential([Dense(1,input_dim = 1)])
```

```python
single.compile(optimizer='adam',
               loss = 'mse'
)
```

```python
history = single.fit(X_train_scaled,
                     y_train,
                     epochs = 500,
                     verbose = 1
                     )
```

```python
single_prediction = single.predict(X_test_scaled)
```

```python
print('r2_score:', r2_score(y_test,single_prediction))
```

```python
multi_layer_model = Sequential([
    Dense(8, activation='relu',input_dim=1),
    Dense(4,activation='relu',input_dim=1),
    Dense(1)
])


multi_layer_model.compile(optimizer='adam',loss='mse')
```

```python
multi_layer_model.fit(X_train_scaled,y_train,epochs = 700,verbose = 1)
multi_prediction = multi_layer_model.predict(X_test_scaled)
```

```python
print('r2_score:',r2_score(y_test,multi_prediction))
```
